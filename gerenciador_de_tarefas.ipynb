```python
def exibir_menu():
    print("\n" + "="*45)
    print("GERENCIADOR DE TAREFAS")
    print("="*45)
    print("1 - Adicionar Nova Tarefa")
    print("2 - Listar Todas as Tarefas")
    print("3 - Concluir uma Tarefa")
    print("4 - Remover uma Tarefa")
    print("5 - Sair do Programa")
    print("="*45)
def main():
    banco_tarefas = []
    id_atual = 1
    while True:
        exibir_menu()
        try:
            opcao = int(input("Escolha uma opção (1 a 5): "))
        except ValueError:
            print("\n [ERRO] Entrada inválida! Por favor, digite apenas números inteiros.")
            continue
        if opcao == 1:
            print("\n --- ADICIONAR TAREFA ---")
            titulo = input("Nome da tarefa/trabalho: ").strip()
            if not titulo:
                print("O nome da tarefa não pode estar vazio!")
                continue
            materia = input("Matéria: ").strip()
            nova_tarefa = {
                "id": id_atual,
                "titulo": titulo,
                "materia": materia if materia else "Geral",
                "concluida": False
            }

            banco_tarefas.append(nova_tarefa)
            print(f"\n Tarefa '{titulo}' adicionada com sucesso. (ID: {id_atual})")
            id_atual += 1

        elif opcao == 2:
            print("\n --- LISTA DE TAREFAS ATIVAS ---")
            if not banco_tarefas:
                print("Nenhuma tarefa cadastrada até o momento.")
                continue

            for tarefa in banco_tarefas:
                id_t = tarefa.get("id", "N/A")
                titulo_t = tarefa.get("titulo", "Sem Título")
                materia_t = tarefa.get("materia", "Geral")
                concluida_t = tarefa.get("concluida", False)

                status = "[CONCLUÍDA]" if concluida_t else "[PENDENTE]"
                print(f"ID {id_t} | {status} - {titulo_t} ({materia_t})")

        elif opcao == 3:
            print("\n --- CONCLUIR TAREFA ---")
            if not banco_tarefas:
                print("Não há tarefas cadastradas para concluir.")
                continue
            try:
                id_busca = int(input("Digite o ID da tarefa que deseja concluir: "))
            except ValueError:
                print("ID inválido! Digite apenas números.")
                continue

            encontrada = False
            for tarefa in banco_tarefas:
                if tarefa.get("id") == id_busca:
                    tarefa["concluida"] = True
                    print(f"\n Parabéns! A tarefa '{tarefa.get('titulo')}' foi marcada como concluída!")
                    encontrada = True
                    break

            if not encontrada:
                print(f" Nenhuma tarefa encontrada com o ID {id_busca}.")

        elif opcao == 4:
            print("\n --- REMOVER TAREFA ---")
            if not banco_tarefas:
                print("Não há tarefas cadastradas para remover.")
                continue

            try:
                id_busca = int(input("Digite o ID da tarefa que deseja remover: "))
            except ValueError:
                print("ID inválido! Digite apenas números.")
                continue

            encontrada = False
            for tarefa in banco_tarefas:
                if tarefa.get("id") == id_busca:
                    banco_tarefas.remove(tarefa)
                    print(f"\n A tarefa '{tarefa.get('titulo')}' foi removida do sistema com sucesso.")
                    encontrada = True
                    break
            if not encontrada:
                print(f"Nenhuma tarefa encontrada com o ID {id_busca}.")
        elif opcao == 5:
            print("\n Encerrando o gerenciador de tarefas acadêmicas.........")
            break
        else:
            print("\n Opção incorreta! Por favor, escolha um número de 1 a 5.")
if __name__ == "__main__":
    main()
```
