```python
import sqlite3

conexao = sqlite3.connect("bloco_notas_.db")
cursor = conexao.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS TAB_NOTAS (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    conteudo TEXT NOT NULL
)
""")
conexao.commit()

while True:
    print("\n==============================================")
    print("--- BLOCO DE NOTAS COM BANCO DE DADOS ---")
    print("==============================================")
    print("1. Adicionar nova anotação (INSERT)")
    print("2. Exibir todas as anotações (SELECT)")
    print("3. Excluir uma anotação específica (DELETE WHERE)")
    print("4. Limpar todas as anotações (DELETE ALL)")
    print("5. Sair do programa")
    print("==============================================")

    opcao = input("Escolha uma opção (1-5): ")

    if opcao == "1":
        nova_nota = input("\nDigite a sua anotação: ")
        if nova_nota.strip() != "":
            cursor.execute("INSERT INTO TAB_NOTAS (conteudo) VALUES (?)", (nova_nota,))
            conexao.commit()
            print("-> Anotação salva com sucesso no Banco de Dados!")
        else:
            print("-> A anotação não pode estar vazia.")

    elif opcao == "2":
        cursor.execute("SELECT id, conteudo FROM TAB_NOTAS")
        anotacoes = cursor.fetchall()

        if len(anotacoes) == 0:
            print("\n[O seu banco de dados não possui nenhuma anotação salva!]")
        else:
            print("\n--- SUAS ANOTAÇÕES SALVAS NO BANCO ---")
            for id_nota, texto in anotacoes:
                print(f"ID: {id_nota} | Nota: {texto}")
            print("=======================================")
            print(f"Total de registros na tabela: {len(anotacoes)}")

    elif opcao == "3":
        cursor.execute("SELECT id, conteudo FROM TAB_NOTAS")
        anotacoes = cursor.fetchall()

        if len(anotacoes) == 0:
            print("\n[Não há registros para excluir!]")
        else:
            print("\n--- SELECIONE O ID QUE DESEJA APAGAR ---")
            for id_nota, texto in anotacoes:
                print(f"[{id_nota}] {texto}")
            print("========================================")

            try:
                id_excluir = int(input("Digite o ID do registro a ser excluído: "))
                cursor.execute("DELETE FROM TAB_NOTAS WHERE id = ?", (id_excluir,))
                conexao.commit()
                print(f"-> Nota com ID {id_excluir} removida com sucesso!")
            except ValueError:
                print("-> Erro: Por favor, digite um número inteiro de ID válido.")

    elif opcao == "4":
        confirmacao = input("\nTem certeza que deseja apagar TUDO do banco? (s/n): ")
        if confirmacao.lower() == "s":
            cursor.execute("DELETE FROM TAB_NOTAS")
            conexao.commit()
            print("-> Banco de dados limpo com sucesso!")
        else:
            print("-> Operação cancelada.")

    elif opcao == "5":
        print("\n-> Salvando alterações pendentes e fechando banco...")
        conexao.close()
        print("-> Conexão encerrada com sucesso...")
        break

    else:
        print("-> Opção incorreta. Escolha um número de 1 a 5.")
```
